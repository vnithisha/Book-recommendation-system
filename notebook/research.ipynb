# Create a pivot table

```python
book_pivot = final_ratings_df.pivot_table(columns='user_id', index='title', values='rating')
```

```python
book_pivot
```

The key distinction:

NaN → no data point exists (absence)

0.0 → a data point exists, and its value happens to be zero (a real recorded "no rating given")

```python
book_pivot.shape
```

### Convert the NaNs to 0s before converting to a sparse matrix

```python
book_pivot.fillna(0, inplace=True)
```

```python
book_pivot
```

```python
from scipy.sparse import csr_matrix
```

```python
book_sparse = csr_matrix(book_pivot)
```

```python
book_sparse
```

Only 14,787 user-book rating pairs actually exist out of a possible 703 × 888 = 624,264 combinations.

```python
sparsity = 1 - (14787 / (703 * 888))
print(f"{sparsity:.2%}")
```

That works out to about 97.6% sparse — meaning only ~2.4% of all possible user-book cells have an actual rating. 
The other 97.6% are zeros (either genuinely unrated, or explicit-0 implicit ratings

```python
type(book_sparse)
```

```python
from sklearn.neighbors import NearestNeighbors
```

```python
book_pivot.index.get_loc("Harry Potter and the Chamber of Secrets (Book 2)")
```

## 1. Test with current Euclidean-based model (already fitted)

```python
# book_pivot.iloc[237,:]
```

```python
model = NearestNeighbors(algorithm='brute', metric='euclidean')
model.fit(book_sparse)

distance, suggestion = model.kneighbors(book_pivot.iloc[224,:].values.reshape(1,-1), n_neighbors=6)

print("Euclidean results:")
for i in suggestion[0]:
    print(book_pivot.index[i])
print("Distances:", distance[0])
```

## 2. Now refit with cosine similarity

```python
model_cosine = NearestNeighbors(algorithm='brute', metric='cosine')
model_cosine.fit(book_sparse)

distance_cos, suggestion_cos = model_cosine.kneighbors(book_pivot.iloc[224,:].values.reshape(1,-1), n_neighbors=6)

print("\nCosine results:")
for i in suggestion_cos[0]:
    print(book_pivot.index[i])
print("Distances:", distance_cos[0])
```

```python
book_pivot.index[224]
```

### Map the neighbor indices to titles (for both models):

```python
print("Queried book:", book_pivot.index[224])

print("\nEuclidean neighbors:")
for i in suggestion[0]:
    print(book_pivot.index[i])

print("\nCosine neighbors:")
for i in suggestion_cos[0]:
    print(book_pivot.index[i])
```

###  To see book + distance together (more useful for comparison):

```python
print("Euclidean:")
for idx, dist in zip(suggestion[0], distance[0]):
    print(f"{book_pivot.index[idx]}  —  distance: {dist:.4f}")

print("\nCosine:")
for idx, dist in zip(suggestion_cos[0], distance_cos[0]):
    print(f"{book_pivot.index[idx]}  —  distance: {dist:.4f}")
```

```python
final_model = NearestNeighbors(algorithm='brute', metric='cosine')
final_model.fit(book_sparse)

distance, suggestion = final_model.kneighbors(book_pivot.iloc[224,:].values.reshape(1,-1), n_neighbors=6)
```
